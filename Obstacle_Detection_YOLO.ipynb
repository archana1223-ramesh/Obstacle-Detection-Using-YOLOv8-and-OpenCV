```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
```

```python
model = YOLO("yolov8n.pt")
```

```python
results = model("classroom.jpg")
```

```python
result = results[0]

output = result.plot()

plt.figure(figsize=(12,8))
plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
results = model("bus.jpg")
```

```python
result = results[0]

output = result.plot()

plt.figure(figsize=(12,8))
plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
results = model.predict(
    source="bike and car.mp4",
    save=True,
    conf=0.3
)
```
